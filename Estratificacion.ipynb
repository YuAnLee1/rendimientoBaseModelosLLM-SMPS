```python
import pandas as pd

df = pd.read_csv("smps_clean_pool.csv")

print("Total de registros cargados:", len(df))
print(df["label_name"].value_counts())
df.head()
```

Se realizará un split 80-10-10 con una semilla fija de valor 42

```python
from sklearn.model_selection import train_test_split

SEED = 42

# Separamos 80% train y el 20% restante será validation + test, estratificado por clase
df_train, df_temp = train_test_split(
    df,
    test_size=0.20,
    stratify=df["label_name"],
    random_state=SEED
)

# Dividimos el 20% restante en mitad validation (10%) mitad test (10%)
df_val, df_test = train_test_split(
    df_temp,
    test_size=0.50,
    stratify=df_temp["label_name"],
    random_state=SEED
)

print("Train:", df_train.shape)
print("Validation:", df_val.shape)
print("Test:", df_test.shape)

for nombre, split_df in {"train": df_train, "validation": df_val, "test": df_test}.items():
    print(f"\n{nombre.upper()}")
    print(split_df["label_name"].value_counts())
    print((split_df["label_name"].value_counts(normalize=True) * 100).round(2))
```

```python
# Guardamos los splits

df_train.to_csv("train_final.csv", index=False)
df_val.to_csv("val_final.csv", index=False)
df_test.to_csv("test_final.csv", index=False)

print("Archivos guardados: train_final.csv, val_final.csv, test_final.csv")
```
